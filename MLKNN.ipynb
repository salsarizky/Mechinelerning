```python
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd
```

```python
data = {
    'Umur (X1)': [40, 50, 50, 70, 80, 60],
    'Credit Rating (X2)': [5, 8, 7, 4, 4, 6],
    'Nilai Pinjaman (X3)': [60, 40, 30, 60, 80, 60],
    'Status': ['TOLAK', 'TERIMA', 'TOLAK', 'TERIMA', 'TERIMA', 'TERIMA']
}

df = pd.DataFrame(data)

print("Data Training")
display(df)
```

```python
X = df[['Umur (X1)', 'Credit Rating (X2)', 'Nilai Pinjaman (X3)']]
y = df['Status']
```

```python
data_baru = [[50, 3, 40]]

print("Data Baru:")
print(data_baru)
```

```python
knn2 = KNeighborsClassifier(n_neighbors=2)
knn2.fit(X, y)

hasil_k2 = knn2.predict(data_baru)

print("\nHasil Prediksi K=2")
print("Status Pinjaman =", hasil_k2[0])
```

```python
knn5 = KNeighborsClassifier(n_neighbors=5)
knn5.fit(X, y)

hasil_k5 = knn5.predict(data_baru)

print("\nHasil Prediksi K=5")
print("Status Pinjaman =", hasil_k5[0])
```

```python
print("\nKesimpulan:")
print("K=2 :", hasil_k2[0])
print("K=5 :", hasil_k5[0])
```
