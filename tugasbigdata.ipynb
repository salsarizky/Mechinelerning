```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
import os

# Check if the file exists, if not, prompt for upload
file_name = 'Crop_recommendation.csv'
file_present = os.path.exists(file_name)

if not file_present:
    print(f"'{file_name}' not found. Please upload the file.")
    uploaded = files.upload()
    if uploaded is not None and file_name in uploaded:
        print(f"'{file_name}' uploaded successfully.")
        file_present = True # Update flag after successful upload
    else:
        print(f"Error: '{file_name}' was not uploaded, or no file was selected. Please ensure you upload the correct file.")
else:
    print(f"'{file_name}' found in the environment.")

# Load dataset and perform analysis only if the file is present
if file_present:
    df = pd.read_csv(file_name)

    # Tampilkan 5 data awal
    print("\nData Awal:")
    print(df.head())

    # Info dataset
    print("\nInfo Dataset:")
    print(df.info())

    # =========================
    # STATISTIK DESKRIPTIF
    # =========================
    print("\nStatistik Deskriptif:")
    print(df.describe())

    # Mean, Median, Mode
    print("\nMean:")
    print(df.mean(numeric_only=True))

    print("\nMedian:")
    print(df.median(numeric_only=True))

    print("\nModus:")
    print(df.mode().iloc[0])

    print("\nVariance:")
    print(df.var(numeric_only=True))

    print("\nStandard Deviation:")
    print(df.std(numeric_only=True))

    # =========================
    # HISTOGRAM
    # =========================
    df.hist(figsize=(12, 8))
    plt.suptitle("Histogram Distribusi Data")
    plt.show()

    # =========================
    # BOXPLOT
    # =========================
    plt.figure()
    df.boxplot()
    plt.title("Boxplot Data")
    plt.xticks(rotation=45)
    plt.show()

    # =========================
    # ANALISIS PER LABEL TANAMAN
    # =========================
    print("\nJumlah tiap tanaman:")
    print(df['label'].value_counts())

    # Grafik jumlah tanaman
    df['label'].value_counts().plot(kind='bar')
    plt.title("Jumlah Data Tiap Tanaman")
    plt.xlabel("Jenis Tanaman")
    plt.ylabel("Jumlah")
    plt.xticks(rotation=90)
    plt.show()
else:
    print("\nDataframe 'df' could not be created because the file was not found or uploaded.")
```
