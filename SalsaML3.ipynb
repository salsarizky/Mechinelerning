```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
import pandas as pd
```

```python
#DATA TRAINING
data = {
    "X1": [40, 50, 50, 70, 80, 60],
    "X2": [5, 8, 7, 4, 4, 6],
    "X3": [60, 40, 30, 60, 80, 60],
    "Y": ["TOLAK", "TERIMA", "TOLAK", "TERIMA", "TERIMA", "TERIMA"]
}

df = pd.DataFrame(data)

print("Data Training:")
display(df)
```

```python
#LABEL
X = df[["X1", "X2", "X3"]].values
y = df["Y"].values
```

```python
#DATA BARU
data_baru = np.array([[50, 3, 40]])
```

```python
#HITUNG JARAK EUCLIDEAN
jarak = np.sqrt(np.sum((X - data_baru) ** 2, axis=1))

hasil_jarak = df.copy()
hasil_jarak["Jarak"] = jarak
hasil_jarak = hasil_jarak.sort_values("Jarak")

print("\nHasil Perhitungan Jarak:")
display(hasil_jarak)
```

```python
#PRREDIKSI K = 2
knn2 = KNeighborsClassifier(n_neighbors=2)
knn2.fit(X, y)
hasil_k2 = knn2.predict(data_baru)
```

```python
#PREDIKSI K = 5
knn5 = KNeighborsClassifier(n_neighbors=5)
knn5.fit(X, y)
hasil_k5 = knn5.predict(data_baru)

print("\nHasil Prediksi K=2")
print("Status Pinjaman =", hasil_k2[0])

print("\nHasil Prediksi K=5")
print("Status Pinjaman =", hasil_k5[0])

print("\nKesimpulan:")
print("K=2 :", hasil_k2[0])
print("K=5 :", hasil_k5[0])
```
